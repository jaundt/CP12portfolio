# 1.1 — Variables & Data Types

Every program starts the same way: giving names to pieces of information so you can use them later. That's all a **variable** is — a labeled box that holds a value.

This lesson is also the first step toward a bigger project: over the next few classes, you'll build a text adventure game, piece by piece, using exactly the concepts we cover along the way. Today's piece is variables — you'll be using them to describe a character before long.

## The Idea

A variable is created with an **assignment statement**: a name, an `=` sign, and a value.

```python
health = 100
```

Python figures out the **data type** automatically, based on the value you give it. There are four you'll use constantly:

| Type | Looks like | Example |
| --- | --- | --- |
| `int` (integer) | a whole number, no quotes | `health = 100` |
| `float` (decimal) | a number with a decimal point | `damage_multiplier = 1.5` |
| `str` (string) | text, wrapped in quotes | `character_name = "Zara"` |
| `bool` (boolean) | exactly `True` or `False`, no quotes | `is_alive = True` |

You can check any variable's type with the built-in `type()` function — that's the fastest way to settle an argument about what something actually is.

````python
# A worked example — describing a game character with all four types

character_name = "Zara"       # str  -> text, in quotes
health = 100                  # int  -> whole number
damage_multiplier = 1.5       # float -> has a decimal point
is_alive = True                # bool -> True or False, no quotes

print(character_name)
print(health)
print(damage_multiplier)
print(is_alive)

print(type(character_name))
print(type(health))
print(type(damage_multiplier))
print(type(is_alive))
````

*Watch out for two easy mistakes:*
- *A string without quotes (`character_name = Zara`) isn't text — Python will think `Zara` is another variable and crash with a `NameError`.*
- *`True` and `False` are capitalized, with no quotes. `true`, `false`, and `"True"` are all something else entirely (and two of them will crash).*
