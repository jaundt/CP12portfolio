# 1.5 — Loops

Every program so far has run each line exactly once, top to bottom. Loops let you repeat a block of code — either a fixed number of times, or until something changes. This is where your programs stop feeling like a list of instructions and start feeling like actual software.

## The Idea

A **`for` loop** repeats a block a set number of times — `range(n)` gives you the numbers `0` through `n-1`:

```python
for i in range(3):
    print(f"Round {i + 1}")
```

A **`while` loop** repeats as long as a condition stays `True` — use it when you don't know in advance how many times you'll need to repeat:

```python
while health > 0:
    health = health - 10
```

`break` exits a loop immediately, even if its condition is still `True` — handy for stopping the moment something happens, rather than waiting for the loop's normal end condition.

````python
# A worked example — Zara explores three rooms in a row

for room_number in range(3):
    print(f"Entering room {room_number + 1}...")

print("All rooms explored.")
````

*Watch out for:* a `while` loop needs something inside it that eventually makes its condition `False` — forget to update the variable it checks, and you've built an infinite loop that never stops. If a cell seems frozen, that's almost always why; use the notebook's stop/interrupt button and check your loop body.

## Guided Practice 1

The monster's health counts down in a `while` loop as Zara lands hits. Run the cell as-is, then change the starting health or the damage per hit and re-run.

````python
# Change the values below and re-run

monster_health = 40
damage_per_hit = 4

while monster_health > 0:
    monster_health = monster_health - damage_per_hit
    print(f"Monster health: {monster_health}")

print("Monster defeated!")
````
